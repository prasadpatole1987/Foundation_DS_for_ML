# What is regression?

**Regression** is simple, common and highly useful data-analysis technique. It is also known as "Fitting a line". Regression fits a straight line between one variable(feature) and a label. It can also find non-linear relationships between a single lable and multiple features.

**Simple linear regression** models a linear relationship between a single feature and a usually continuous label, allowing the feature to predict the label.
Simple linear regression has two parameters: an **intercept (c)**, which indicates the value that the label is when the feature is set to zero; and a **slope(m)** which indicates how much the label will increase for each one-point increase in the feature. 

y = mx + c
where, y is label and x is your feature. 

**Advantages**: 
**Predictable and easy to interpret** Regressions are easy to interpret because they describe simple mathematical equations. 

**Easy to extrapolate** regression make it easy to extrapolate; to make predictions for values outside the range of our dataset. 

**Optimal fitting is usually guaranteed** Linear regression does not need special attention to be paid to the fitting process, and the optimal solution is guranteed as long as sample size is not too long.

```python
import pandas as pd

dataset = pd.read_csv('doggy-illness.csv', delimiter="\t")
dataset.head()
```

```python
# Data visualization

import matplotlib.pyplot as plt

# Histogram for 'age'
plt.figure(figsize=(10,5))
plt.hist(dataset['age'], bins=10, edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Feature')
plt.show()

# Histogram for 'core_temperature'
plt.figure(figsize=(10,5))
plt.hist(dataset.core_temperature, bins=10, edgecolor='black')
plt.xlabel('Core Temperature')
plt.ylabel('Frequency')
plt.title('Label')
plt.show()
```

```python
import matplotlib.pyplot as plt

# scatter plot for 'age' vs 'core_temperature'
plt.figure(figsize=(10,5))
plt.scatter(dataset.age, dataset.core_temperature, edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Core Temperature')
plt.title('Core temperature vs age')
plt.show()
```

```python
# Simple linear regression

import statsmodels.formula.api as smf
import matplotlib.pyplot as plt

formula = 'core_temperature ~ age'

model = smf.ols(formula=formula, data= dataset).fit()

print(f'slope = {model.params.iloc[1]}, intercept: {model.params.iloc[0]}')

plt.figure(figsize=(10, 5))
plt.scatter(dataset.age, dataset.core_temperature, edgecolor='black')
plt.plot(dataset.age, model.params.iloc[1] * dataset.age + model.params.iloc[0], color='red')
plt.xlabel('Age')
plt.ylabel('Core Temperature')
plt.title('Core temperature as function of a age')
plt.show()
```

```python
def estimate_temperature(age, model):
    return age * model.params.iloc[1] + model.params.iloc[0]

print('Estimate temperature from age')
print(estimate_temperature(10,model))
```
