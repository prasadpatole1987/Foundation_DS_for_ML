# What is Pandas?

### Pandas is an extremely popular Python library for data analysis and manipulation.

### Data exploration and data analysis is typically an iterative process, in which the data scientist takes a sample of data and performs the following kinds of tasks to analyze it and test hypothesis.

<ol>
<li> <strong>Clean Data</strong> to handle errors, missing values and other issues. </li>
<li><strong>Apply statistical techniques to better understand the data</strong> and how the sample might be expected to represent the real-world population of data, allowing for random variation.</li>
<li><strong>Visualize data</strong> to determine relationships between variables and in the case of a machine learning project identify features that are potentially predictive of the labels.</li>
<li><strong>Revise the hypothesis</strong> and repeat the process.</li>
</ol>

```python
### Numpy arrays

data = [50,50,47,97,49,3,53,42,26,74,82,62,37,15,70,27,36,35,48,52,63,64]
print(data)
```

```python
import numpy as np

grades = np.array(data)
print(grades)

print(type(data), 'x 2', data * 2)

print('------')

print(type(grades), 'x 2', grades * 2)
```

```python
print(grades.shape)

print(grades[0])

print(grades.mean())
```

```python
# Define an array of study hours
study_hours = [10.0,11.5,9.0,16.0,9.25,1.0,11.5,9.0,8.5,14.5,15.5,13.75,9.0,8.0,15.5,8.0,9.0,6.0,10.0,12.0,12.5,12.0]


# Create a 2D array (an array of arrays)
student_data = np.array([study_hours, grades])


# display the array
student_data
```

```python
print(student_data.shape)

print(student_data[0][0])

avg_study = student_data[0].mean()

avg_grade = student_data[1].mean()

print(f'Average study hours: { avg_study } \nAverage grade: { avg_grade }')
```

```python
### Exploring tabular data with Pandas 

import pandas as pd

df_students = pd.DataFrame({'Name': ['Dan', 'Joann', 'Pedro', 'Rosie', 'Ethan', 'Vicky', 'Frederic', 'Jimmie', 'Rhonda', 'Giovanni', 'Francesca', 'Rajab', 'Naiyana', 'Kian', 'Jenny', 'Jakeem','Helena','Ismat','Anila','Skye','Daniel','Aisha'], 
                            'StudyHours': study_hours, 
                            'Grade': grades})


df_students
```

```python
# Finding and filtering data in DataFrames

print(df_students.loc[5])

print(df_students.loc[0:5])

print(df_students.iloc[0:5])

print(df_students.iloc[0, [1,2]])

print(df_students.loc[0, 'Grade'])

print(df_students.loc[df_students['Name'] == 'Aisha'])

print(df_students[df_students['Name'] == 'Aisha'])

print(df_students.query("Name == 'Aisha'"))
```

```python
# Handling missing values 

print(df_students.isnull())

print(df_students.isnull().sum())

df_students[df_students.isnull().any(axis=1)]
```

```python
df_students.StudyHours = df_students.StudyHours.fillna(df_students.StudyHours.mean())

df_students
```

```python
df_students = df_students.dropna(axis=0, how='any')
df_students
```

```python
# Explore data in the DataFrame

mean_study = df_students['StudyHours'].mean()

mean_grade = df_students['Grade'].mean()

print(f'Average study hours: { mean_study } \nAverage grade: { mean_grade }')
```

```python
print(df_students[df_students.StudyHours > mean_study])

print(df_students[df_students.Grade > mean_grade].Grade.mean())
```

```python
passes = pd.Series(df_students.Grade >= 60)

df_students = pd.concat([df_students, passes.rename('Pass')], axis=1)

df_students
```

```python
df_students = df_students.sort_values(by='Grade', ascending=False)
df_students
```
