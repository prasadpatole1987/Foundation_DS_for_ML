**Features** is an individual, measurable property or characteristic of a data point.
Features are the input to the model which are used to learn patterns and make decisions.

**Labels** are the outcome or target variable that a machine learning model is trained to predict.
Labels guides the model to the correct answer.

**R2 Score**: It measures the proportion of the variance in a dependent variable that is predictable from an independent variable or set of independent variables in a regression model.

It tells us 
**Goodness of fit**: It Provides a measure of how well the regression model fits the data. An R2 of 1.0 means the model perfectly predicts the outcome. While R2 of 0.0 means it does not predict the outcome any better than simply using average.

**Predictive Power**: R2 score indicates how much of the variability in the dependent variable can be explained by the independent variables. 
For example, an R2 score of 0.75 means that 75% of the total variation in the dependent variable can be explained by the model. 

**Model comparison** R2 value can be used to compare different regression models. A model with a higher R2 is generally considered to be a better fit.

**Scale independent** It is a normalized measure, making it useful for comparing the performance of models across different datasets that may have different scales or units.

**Mean Squared Error(MSE)** is a metric that measures the average of the squares of the errors, or difference between the predicted and actual values. It is common way to predict the performance of regression models. 

It provides single number to quantify the overall error of a model.
A lower MSE indicates that the model is more accurate.
It is often used as a loss function during model training to help model learn by minimizing the error. 



# Regression algorithms

## Simple Linear Regression
Models the relationship between a single independetn variable and a dependent variable using a straight line.


## Multiple Linear Regression
Extends the simple linear regression to model the relationship between two or more independent variables and a dependent variable. 


**Polynomial Regression**
Used when the relationship between variables is non-linear. It fits a curved line to the data by including polynomial terms of the independent variables.

**Ridge Regression**
A type of linear regression that includes an L2 regularization term to prevent overfitting by penelizing large co-efficients.

**Lasso Regression**
Another type of linear regression with an L1 regularization term. It can perform feature selection by shrinking some coefficients to exactly zero. 

**Decision Tree Regression**
Uses a tree like structure to make predictions by splitting the data based on features until a stopping criterion is met. 

**Random Forest Regression**
An ensemble method that builds multiple decision trees and combines their prediction to improve accuracy and reduce overfitting.

**Support Vector Regression**
An extension of support vector machines for regression tasks. It aims to find a function that deviates from the actual target values by a certain margin. 

**Gradient Boosting Regression (e.g. Gradient Boosting Machines, XGBoost, LightGBM)**
Ensemble methods that builds a sequence of weak learners where each subsequent learner corrects the errors of previous ones. 

**Bayesian Linear Regression**
Incorporates prior knowledge about the model parameters into the regression analusis, providing a probabilistic interpretation of the predictions.
